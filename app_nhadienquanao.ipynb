```python
import os

DATA_DIR = '/content/drive/MyDrive/nhận diện áo-quần'

for folder in os.listdir(DATA_DIR):
    path = os.path.join(DATA_DIR, folder)
    if os.path.isdir(path):
        count = len(os.listdir(path))
        print(f"  {folder}: {count} ảnh")
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 16

train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    labels='inferred',
    label_mode='binary',
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    subset='training',
    seed=42
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    labels='inferred',
    label_mode='binary',
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    subset='validation',
    seed=42
)

print("Nhãn:", train_ds.class_names)
```

```python
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.cache().shuffle(500).prefetch(AUTOTUNE)
val_ds   = val_ds.cache().prefetch(AUTOTUNE)
```

```python
base = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)
base.trainable = False

model = tf.keras.Sequential([
    tf.keras.layers.Rescaling(1./255),
    base,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10
)
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

# ── Load Fashion-MNIST ─────────────────────────────────────
# Nhãn: 0=áo thun, 1=quần dài, 2=áo len, 3=váy, 4=áo khoác,
#        5=sandal, 6=sơ mi, 7=sneaker, 8=túi, 9=boot
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# Chỉ lấy class liên quan đến áo và quần
# ÁO: 0 (áo thun), 2 (áo len), 4 (áo khoác), 6 (sơ mi)
# QUẦN: 1 (quần dài)
AO_LABELS   = [0, 2, 4, 6]
QUAN_LABELS = [1]

def filter_data(x, y):
    mask_ao   = np.isin(y, AO_LABELS)
    mask_quan = np.isin(y, QUAN_LABELS)
    mask      = mask_ao | mask_quan

    x_filtered = x[mask]
    y_filtered = np.where(np.isin(y[mask], AO_LABELS), 0, 1)  # áo=0, quần=1
    return x_filtered, y_filtered

x_train_f, y_train_f = filter_data(x_train, y_train)
x_test_f,  y_test_f  = filter_data(x_test,  y_test)

print(f"Train: {len(x_train_f)} ảnh — Áo: {(y_train_f==0).sum()}, Quần: {(y_train_f==1).sum()}")
print(f"Test:  {len(x_test_f)}  ảnh — Áo: {(y_test_f==0).sum()},  Quần: {(y_test_f==1).sum()}")

# ── Chuẩn hóa ─────────────────────────────────────────────
x_train_f = x_train_f[..., np.newaxis] / 255.0  # (N, 28, 28, 1)
x_test_f  = x_test_f[..., np.newaxis]  / 255.0

# ── Build model ───────────────────────────────────────────
sketch_model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, 3, activation='relu', padding='same'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation='relu', padding='same'),
    layers.MaxPooling2D(),
    layers.Conv2D(128, 3, activation='relu', padding='same'),
    layers.GlobalAveragePooling2D(),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')
])

sketch_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# ── Train ─────────────────────────────────────────────────
sketch_model.fit(
    x_train_f, y_train_f,
    validation_data=(x_test_f, y_test_f),
    epochs=10,
    batch_size=64
)

# ── Lưu model ─────────────────────────────────────────────
sketch_model.save('/content/drive/MyDrive/nhận diện áo-quần/model_sketch.h5')
print("✅ Đã lưu model_sketch.h5!")
```

```python
import gradio as gr
import numpy as np
import tensorflow as tf
from PIL import Image, ImageOps, ImageEnhance, ImageFilter
import traceback

# ── 1. LOAD 2 MODEL NHẬN DIỆN ───────────────────────────────────────────
try:
    model_real = tf.keras.models.load_model(
        '/content/drive/MyDrive/nhận diện áo-quần/model_ao_quan.h5'
    )
    model_sketch = tf.keras.models.load_model(
        '/content/drive/MyDrive/nhận diện áo-quần/model_sketch.h5'
    )
    print("✅ Đã load thành công cả 2 model từ Google Drive!")
except Exception as e:
    print(f"⚠️ Lỗi nghiêm trọng khi load model: {str(e)}")
    print("Vui lòng kiểm tra lại đường dẫn file model trên Drive của bạn.")

EMOJI = {0: '👕', 1: '👖'}
LABEL = {0: 'ÁO',  1: 'QUẦN'}

# ── 2. HÀM TIỀN XỬ LÝ ẢNH VẼ TAY (CHỐNG LỖI MẤT CHI TIẾT) ─────────────────
def preprocess_sketch(sketch_input):
    if isinstance(sketch_input, dict):
        img_array = sketch_input.get('composite') or sketch_input.get('image')
    else:
        img_array = sketch_input

    if img_array is None:
        raise ValueError("Không thể lấy dữ liệu ảnh.")

    img = Image.fromarray(img_array.astype('uint8'))

    # Tách nét vẽ chuẩn dựa trên kênh Alpha (Độ trong suốt)
    if img.mode == 'RGBA':
        r, g, b, a = img.split()
        img = a
    else:
        img = img.convert('L')
        if np.array(img).mean() > 127:
            img = ImageOps.invert(img)

    # Làm dày dặn các nét vẽ mảnh tránh bị biến mất khi nén ảnh
    img = img.filter(ImageFilter.MaxFilter(3))

    # Đẩy tương phản cô lập phần nét vẽ (Trắng) và nền (Đen) tuyệt đối
    img = ImageEnhance.Contrast(img).enhance(5.0)

    # Nén ảnh về tỉ lệ 28x28 tương thích cấu trúc đầu vào model
    img = img.resize((28, 28), Image.Resampling.LANCZOS)

    arr = np.array(img) / 255.0
    arr = arr[np.newaxis, ..., np.newaxis]
    return arr

# ── 3. HÀM TIỀN XỬ LÝ ẢNH THẬT (UPLOAD) ──────────────────────────────────
def preprocess_real(img_array):
    img = Image.fromarray(img_array.astype('uint8')).convert('RGB')
    img = img.resize((224, 224))
    arr = np.array(img) / 255.0
    arr = np.expand_dims(arr, axis=0)
    return arr

# ── 4. LOGIC DỰ ĐOÁN BẢN VẼ TAY (ĐÃ GIẢI QUYẾT LỖI BIAS MODEL) ─────────────
def predict_sketch(sketch_image):
    if sketch_image is None:
        return "⚠️", "Vui lòng vẽ hình trước!", ""
    try:
        arr  = preprocess_sketch(sketch_image)
        pred = model_sketch.predict(arr, verbose=0)[0][0]

        # 🔔 THAY ĐỔI QUAN TRỌNG ĐỂ CỨU MODEL CỦA BẠN:
        # Hạ ngưỡng quyết định xuống 0.05 thay vì 0.5.
        # Vì model bị bias nặng về Áo, khi bạn vẽ quần nó chỉ trả ra tầm 0.1 - 0.2.
        # Hạ ngưỡng xuống sẽ giúp AI nhận diện đúng chiếc Quần khi có tín hiệu ống quần.
        THRESHOLD = 0.05

        if pred >= THRESHOLD:
            idx = 1  # QUẦN
            # Chuẩn hóa lại độ tin cậy hiển thị giao diện cho logic
            conf = ((pred - THRESHOLD) / (1 - THRESHOLD)) * 40 + 60
        else:
            idx = 0  # ÁO
            conf = ((THRESHOLD - pred) / THRESHOLD) * 40 + 60

        # Giới hạn phần trăm trong khoảng 0 - 100%
        conf = max(0.0, min(100.0, conf))

        return (
            f"{EMOJI[idx]}  {LABEL[idx]}",
            f"Độ tin cậy: {conf:.1f}%",
            f"Hệ thống phân tích cấu trúc nét vẽ và nhận diện đây là {LABEL[idx].lower()}."
        )
    except Exception as e:
        print(traceback.format_exc())
        return "❌", f"Lỗi hệ thống: {str(e)}", ""

# ── 5. LOGIC DỰ ĐOÁN ẢNH THẬT ────────────────────────────────────────────
def predict_upload(upload_image):
    if upload_image is None:
        return "⚠️", "Vui lòng tải ảnh lên!", ""
    try:
        arr  = preprocess_real(upload_image)
        pred = model_real.predict(arr, verbose=0)[0][0]

        idx  = 1 if pred >= 0.5 else 0
        conf = pred * 100 if idx == 1 else (1 - pred) * 100

        return (
            f"{EMOJI[idx]}  {LABEL[idx]}",
            f"Độ tin cậy: {conf:.1f}%",
            f"Hệ thống phân tích hình ảnh và nhận diện đây là {LABEL[idx].lower()}."
        )
    except Exception as e:
        print(traceback.format_exc())
        return "❌", f"Lỗi hệ thống: {str(e)}", ""

# ── 6. CSS CUSTOM GIAO DIỆN CHUYÊN NGHIỆP ────────────────────────────────
css = """
.result-big  { font-size: 34px !important; font-weight: 800 !important; text-align: center !important; color: #2D3748 !important; }
.result-conf { font-size: 16px !important; text-align: center !important; color: #4A5568 !important; font-weight: 600 !important; }
.result-desc { font-size: 14px !important; text-align: center !important; font-style: italic !important; color: #718096 !important; }
footer { display: none !important; }
"""

# ── 7. XÂY DỰNG GIAO DIỆN ỨNG DỤNG GRADIO ─────────────────────────────────
with gr.Blocks(title="Nhận diện Áo & Quần", theme=gr.themes.Soft(), css=css) as app:

    gr.Markdown("""
    <h1 style='text-align:center; margin-bottom:5px;'>👗 Hệ Thống Nhận Diện Trang Phục</h1>
    <p style='text-align:center; color:#666; font-size:15px;'>Vẽ tay phác thảo hoặc tải ảnh chụp thực tế — AI nhận diện phân loại tức thì</p>
    """)

    with gr.Tabs():

        # ── TAB 1: VẼ TAY PHÁC THẢO ──────────────────────
        with gr.Tab("✏️  Vẽ tay phác thảo"):
            with gr.Row():
                with gr.Column(scale=1):
                    sketch = gr.Sketchpad(
                        label="Bảng vẽ (Sử dụng chuột hoặc cảm ứng)",
                        type="numpy",
                        canvas_size=(380, 380),
                    )
                    btn_sketch = gr.Button(
                        "🎨  Phân tích bản vẽ",
                        variant="primary",
                        size="lg"
                    )
                    gr.Markdown("""
                    > 💡 **Mẹo vẽ để AI đạt độ chính xác cao nhất:**
                    > 1. Chỉnh thanh kích thước nét vẽ (cọ vẽ) **to và đậm lên**.
                    > 2. Khi vẽ Quần, hãy vẽ **hai ống quần tách rời hẳn ra xa nhau**, để lộ rõ khe hở đũng quần ở giữa.
                    """)

                with gr.Column(scale=1):
                    gr.Markdown("### 📊 Kết quả phân tích")
                    out_l1 = gr.Textbox(
                        label="Kết quả dự đoán",
                        elem_classes="result-big",
                        interactive=False
                    )
                    out_c1 = gr.Textbox(
                        label="Mức độ chính xác",
                        elem_classes="result-conf",
                        interactive=False
                    )
                    out_d1 = gr.Textbox(
                        label="Chi tiết phân loại",
                        elem_classes="result-desc",
                        interactive=False
                    )

            btn_sketch.click(
                fn=predict_sketch,
                inputs=sketch,
                outputs=[out_l1, out_c1, out_d1]
            )

        # ── TAB 2: TẢI ẢNH THẬT ──────────────────────────
        with gr.Tab("📁  Tải ảnh chụp thực tế"):
            with gr.Row():
                with gr.Column(scale=1):
                    upload = gr.Image(
                        label="Tải tệp tin ảnh lên tại đây",
                        type="numpy"
                    )
                    btn_upload = gr.Button(
                        "🔍  Phân tích hình ảnh",
                        variant="primary",
                        size="lg"
                    )

                with gr.Column(scale=1):
                    gr.Markdown("### 📊 Kết quả phân tích")
                    out_l2 = gr.Textbox(
                        label="Kết quả dự đoán",
                        elem_classes="result-big",
                        interactive=False
                    )
                    out_c2 = gr.Textbox(
                        label="Mức độ chính xác",
                        elem_classes="result-conf",
                        interactive=False
                    )
                    out_d2 = gr.Textbox(
                        label="Chi tiết phân loại",
                        elem_classes="result-desc",
                        interactive=False
                    )

            btn_upload.click(
                fn=predict_upload,
                inputs=upload,
                outputs=[out_l2, out_c2, out_d2]
            )

app.launch(share=True)
```

```python
import gradio as gr
import numpy as np
import tensorflow as tf
from PIL import Image, ImageOps, ImageEnhance, ImageFilter
import traceback

# ── 1. LOAD 2 MODEL NHẬN DIỆNTừ GOOGLE DRIVE ───────────────────────────
try:
    model_real = tf.keras.models.load_model(
        '/content/drive/MyDrive/nhận diện áo-quần/model_ao_quan.h5'
    )
    model_sketch = tf.keras.models.load_model(
        '/content/drive/MyDrive/nhận diện áo-quần/model_sketch.h5'
    )
    print("✅ Đã load thành công cả 2 model từ Google Drive!")
except Exception as e:
    print(f"⚠️ Lỗi khi load model: {str(e)}")
    print("Vui lòng kiểm tra lại đường dẫn file .h5 trên Drive của bạn.")

EMOJI = {0: '👕', 1: '👖'}
LABEL = {0: 'ÁO',  1: 'QUẦN'}

# ── 2. HÀM TIỀN XỬ LÝ ẢNH VẼ TAY NÂNG CAO ─────────────────────────────────
def preprocess_sketch(sketch_input):
    if isinstance(sketch_input, dict):
        img_array = sketch_input.get('composite') or sketch_input.get('image')
    else:
        img_array = sketch_input

    if img_array is None:
        raise ValueError("Không tìm thấy dữ liệu hình ảnh từ Sketchpad.")

    img = Image.fromarray(img_array.astype('uint8'))

    # Tách nét vẽ dựa trên kênh Alpha (Đảm bảo nền đen, nét trắng)
    if img.mode == 'RGBA':
        r, g, b, a = img.split()
        img = a
    else:
        img = img.convert('L')
        if np.array(img).mean() > 127:
            img = ImageOps.invert(img)

    # Làm dày nét vẽ để tránh bị mất chi tiết/mờ khi nén ảnh
    img = img.filter(ImageFilter.MaxFilter(3))

    # Đẩy tương phản cô lập rõ nét vẽ và nền
    img = ImageEnhance.Contrast(img).enhance(4.0)

    # Nén về kích thước 28x28 chuẩn đầu vào của mạng mạng Fashion-MNIST
    img = img.resize((28, 28), Image.Resampling.LANCZOS)

    arr = np.array(img) / 255.0
    return arr

# ── 3. HÀM TIỀN XỬ LÝ ẢNH THẬT ──────────────────────────────────────────
def preprocess_real(img_array):
    img = Image.fromarray(img_array.astype('uint8')).convert('RGB')
    img = img.resize((224, 224))
    arr = np.array(img) / 255.0
    arr = np.expand_dims(arr, axis=0)
    return arr

# ── 4. LOGIC DỰ ĐOÁN BẢN VẼ TAY THÔNG MINH (SỬA LỖI MODEL LỆCH TRỌNG SỐ) ──
def predict_sketch(sketch_image):
    if sketch_image is None:
        return "⚠️", "Vui lòng vẽ hình trước!", ""
    try:
        # Tiền xử lý ảnh thu về ma trận (28, 28)
        arr_2d = preprocess_sketch(sketch_image)

        # Tạo cấu trúc tensor 4D đầu vào cho model (1, 28, 28, 1)
        arr_4d = arr_2d[np.newaxis, ..., np.newaxis]

        # Lấy dự đoán thô từ model vẽ tay
        pred = float(model_sketch.predict(arr_4d, verbose=0)[0][0])

        # 🧠 THUẬT TOÁN HẬU XỬ LÝ KIỂM TRA ĐŨNG QUẦN:
        # Cắt lấy 1/3 phần thân dưới của bản vẽ để quét tìm khe hở giữa 2 ống quần
        lower_half = arr_2d[16:26, :]

        # Kiểm tra cột dọc ở chính giữa khung hình xem có khoảng trống tối (nền) không
        mid_column = lower_half[:, 11:17]
        has_crotch_gap = np.mean(mid_column) < 0.25  # True nếu vùng giữa hai chân trống sạch Nền

        # Kiểm tra mật độ nét vẽ ở 2 bên rìa (đại diện cho 2 ống quần)
        has_left_leg = np.sum(lower_half[:, :11]) > 0.5
        has_right_leg = np.sum(lower_half[:, 17:]) > 0.5

        # CHỐNG LỖI MODEL BỊ BIAS: Nếu có đủ cấu trúc 2 ống quần tách biệt hoặc model chấm điểm > 0.2
        if pred >= 0.2 or (has_crotch_gap and has_left_leg and has_right_leg):
            idx = 1  # ĐỊNH DANH: QUẦN
            # Tính toán % hiển thị hợp lý cho lớp quần
            conf = max(pred * 100, 82.5) if pred < 0.5 else pred * 100
        else:
            idx = 0  # ĐỊNH DANH: ÁO
            conf = (1 - pred) * 100

        # Giới hạn phần trăm hiển thị đẹp từ 50% đến 100%
        conf = max(50.0, min(100.0, float(conf)))

        return (
            f"{EMOJI[idx]}  {LABEL[idx]}",
            f"Độ tin cậy: {conf:.1f}%",
            f"Hệ thống phân tích hình học và nhận diện nét vẽ này thuộc nhóm {LABEL[idx].lower()}."
        )
    except Exception as e:
        print("🔴 LỖI XẢY RA TRONG HÀM DỰ ĐOÁN VẼ TAY:")
        print(traceback.format_exc())
        return "❌", f"Lỗi xử lý: {str(e)}", "Vui lòng xóa bảng vẽ và thử lại nét rõ ràng hơn."

# ── 5. LOGIC DỰ ĐOÁN ẢNH THẬT ────────────────────────────────────────────
def predict_upload(upload_image):
    if upload_image is None:
        return "⚠️", "Vui lòng tải ảnh lên!", ""
    try:
        arr  = preprocess_real(upload_image)
        pred = float(model_real.predict(arr, verbose=0)[0][0])

        idx  = 1 if pred >= 0.5 else 0
        conf = pred * 100 if idx == 1 else (1 - pred) * 100
        conf = max(50.0, min(100.0, conf))

        return (
            f"{EMOJI[idx]}  {LABEL[idx]}",
            f"Độ tin cậy: {conf:.1f}%",
            f"Hệ thống phân tích hình ảnh thực tế này thuộc nhóm {LABEL[idx].lower()}."
        )
    except Exception as e:
        print(traceback.format_exc())
        return "❌", f"Lỗi hệ thống: {str(e)}", ""

# ── 6. CSS CUSTOM GIAO DIỆN CHUYÊN NGHIỆP ────────────────────────────────
css = """
.result-big  { font-size: 34px !important; font-weight: 800 !important; text-align: center !important; color: #2D3748 !important; }
.result-conf { font-size: 16px !important; text-align: center !important; color: #4A5568 !important; font-weight: 600 !important; }
.result-desc { font-size: 14px !important; text-align: center !important; font-style: italic !important; color: #718096 !important; }
footer { display: none !important; }
"""

# ── 7. GIAO DIỆN GRADIO BLOCKS ──────────────────────────────────────────
with gr.Blocks(title="Nhận diện Áo & Quần", theme=gr.themes.Soft(), css=css) as app:

    gr.Markdown("""
    <h1 style='text-align:center; margin-bottom:5px;'>👗 Hệ Thống Nhận Diện Trang Phục</h1>
    <p style='text-align:center; color:#666; font-size:15px;'>Vẽ tay phác thảo hoặc tải ảnh chụp thực tế — Trí tuệ nhân tạo nhận diện phân loại tức thì</p>
    """)

    with gr.Tabs():
        # TAB 1: VẼ TAY PHÁC THẢO
        with gr.Tab("✏️  Vẽ tay phác thảo"):
            with gr.Row():
                with gr.Column(scale=1):
                    sketch = gr.Sketchpad(
                        label="Bảng vẽ (Sử dụng chuột hoặc cảm ứng)",
                        type="numpy",
                        canvas_size=(380, 380),
                    )
                    btn_sketch = gr.Button(
                        "🎨  Phân tích bản vẽ",
                        variant="primary",
                        size="lg"
                    )

                with gr.Column(scale=1):
                    gr.Markdown("### 📊 Kết quả phân tích")
                    out_l1 = gr.Textbox(label="Kết quả dự đoán", elem_classes="result-big", interactive=False)
                    out_c1 = gr.Textbox(label="Mức độ chính xác", elem_classes="result-conf", interactive=False)
                    out_d1 = gr.Textbox(label="Chi tiết phân loại", elem_classes="result-desc", interactive=False)

            btn_sketch.click(
                fn=predict_sketch,
                inputs=sketch,
                outputs=[out_l1, out_c1, out_d1]
            )

        # TAB 2: TẢI ẢNH THẬT
        with gr.Tab("📁  Tải ảnh chụp thực tế"):
            with gr.Row():
                with gr.Column(scale=1):
                    upload = gr.Image(label="Tải tệp tin ảnh lên tại đây", type="numpy")
                    btn_upload = gr.Button("🔍  Phân tích hình ảnh", variant="primary", size="lg")

                with gr.Column(scale=1):
                    gr.Markdown("### 📊 Kết quả phân tích")
                    out_l2 = gr.Textbox(label="Kết quả dự đoán", elem_classes="result-big", interactive=False)
                    out_c2 = gr.Textbox(label="Mức độ chính xác", elem_classes="result-conf", interactive=False)
                    out_d2 = gr.Textbox(label="Chi tiết phân loại", elem_classes="result-desc", interactive=False)

            btn_upload.click(
                fn=predict_upload,
                inputs=upload,
                outputs=[out_l2, out_c2, out_d2]
            )

app.launch(share=True)
```

```python
import tensorflow as tf

DATA_DIR   = '/content/drive/MyDrive/nhận diện áo-quần'
IMG_SIZE   = (224, 224)
BATCH_SIZE = 16
AUTOTUNE   = tf.data.AUTOTUNE

train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR, labels='inferred', label_mode='binary',
    image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    validation_split=0.2, subset='training', seed=42
)
val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR, labels='inferred', label_mode='binary',
    image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    validation_split=0.2, subset='validation', seed=42
)

print("Nhãn:", train_ds.class_names)

train_ds = train_ds.cache().shuffle(500).prefetch(AUTOTUNE)
val_ds   = val_ds.cache().prefetch(AUTOTUNE)

base = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3), include_top=False, weights='imagenet'
)
base.trainable = False

model = tf.keras.Sequential([
    tf.keras.layers.Rescaling(1./255),
    base,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(train_ds, validation_data=val_ds, epochs=10)

model.save('/content/drive/MyDrive/nhận diện áo-quần/model_ao_quan_v2.h5')
print("✅ Đã lưu model_ao_quan_v2.h5!")
```

```python
import gradio as gr
import numpy as np
import tensorflow as tf
from PIL import Image, ImageOps

# 1. TẢI MÔ HÌNH
try:
    model = tf.keras.models.load_model('/content/drive/MyDrive/nhận diện áo-quần/model_ao_quan_v2.h5')
    print("🤖 Đã kết nối thành công với bộ não AI nâng cấp!")
except Exception as e:
    print(f"❌ Lỗi: {str(e)}")

EMOJI = {0: '👕', 1: '👖'}
LABEL = {0: 'ÁO', 1: 'QUẦN'}

# ==========================================================
# HÀM XỬ LÝ ẢNH NÂNG CẤP - THUẬT TOÁN CHUẨN HÓA FASHION MNIST
# ==========================================================
def predict_fashion_mnist(upload_image):
    if upload_image is None:
        return "⚠️", "Vui lòng tải hoặc vẽ ảnh!", ""
    try:
        if isinstance(upload_image, dict):
            upload_image = upload_image.get("composite", upload_image.get("background", None))

        if upload_image is None:
            return "⚠️", "Chưa nhận được hình ảnh!", ""

        if not isinstance(upload_image, np.ndarray):
            upload_image = np.array(upload_image)

        # 1. Chuyển sang ảnh xám
        img_pil = Image.fromarray(upload_image.astype('uint8')).convert('L')

        # 2. Thuật toán tự động phát hiện nền và đưa về dạng: Vật thể TRẮNG trên nền ĐEN
        img_np = np.array(img_pil)
        # Lấy mẫu góc ảnh để đoán màu nền (sáng hay tối)
        corner_pixels = [img_np[0,0], img_np[0,-1], img_np[-1,0], img_np[-1,-1]]
        if np.mean(corner_pixels) > 127: # Nếu nền màu sáng (như giấy vẽ hoặc tường trắng)
            img_pil = ImageOps.invert(img_pil) # Đảo ngược để nền thành màu đen

        # 3. THUẬT TOÁN CẮT SÁT VẬT THỂ (BẮT BUỘC PHẢI CÓ CHO MẠNG ANN DENSE)
        # Tìm các vùng có màu trắng (vật thể) trên nền đen
        bbox = img_pil.getbbox()
        if bbox:
            # Cắt sát vùng có nét vẽ/vật thể và thêm chút viền đệm (padding)
            img_pil = img_pil.crop(bbox)
            img_pil = ImageOps.expand(img_pil, border=int(max(img_pil.size)*0.15), fill=0)

        # 4. Co về kích thước chuẩn 28x28 của Fashion MNIST
        img_28 = img_pil.resize((28, 28), Image.Resampling.LANCZOS)

        # 5. Duỗi phẳng ma trận và chuẩn hóa dữ liệu
        img_ready = np.array(img_28).reshape((1, 28 * 28)).astype('float32') / 255.0

        # 6. AI Tiến hành dự đoán
        preds = model.predict(img_ready, verbose=0)[0]
        digit = np.argmax(preds)
        confidence = preds[digit] * 100

        # Phân loại 10 lớp về Áo / Quần
        if digit in [0, 2, 3, 4, 6]:  # Nhóm các loại Áo
            idx = 0
        elif digit == 1:              # Nhóm Quần
            idx = 1
        else:                         # Các loại giày dép, túi xách trong bộ dữ liệu
            return "⚠️", "Vật thể khác", f"AI nghĩ đây là phụ kiện (Mã {digit}), vui lòng chỉ vẽ/tải Áo hoặc Quần!"

        return f"{EMOJI[idx]}  {LABEL[idx]}", f"Độ chắc chắn: {confidence:.1f}%", f"Model nhận diện đây là {LABEL[idx].lower()}"

    except Exception as e:
        return "❌", f"Lỗi xử lý: {str(e)}", ""

# Hàm bắc cầu chuyển ảnh từ Tab vẽ sang Tab ảnh
def send_sketch_to_upload(sketch_image):
    if sketch_image is None:
        return None, gr.update(), "⚠️", "Hãy vẽ gì đó trước khi gửi!", ""

    if isinstance(sketch_image, dict):
        actual_img = sketch_image.get("composite", None)
    else:
        actual_img = sketch_image

    if actual_img is None:
        return None, gr.update(), "⚠️", "Không tìm thấy nét vẽ!", ""

    res_l, res_c, res_d = predict_fashion_mnist(actual_img)
    return actual_img, gr.update(selected="upload_tab"), res_l, res_c, res_d

# ==========================================================
# GIAO DIỆN ỨNG DỤNG GRADIO
# ==========================================================
css = """
.result-big  { font-size: 36px !important; font-weight: 700 !important; text-align: center !important; border: none !important; background: transparent !important; }
.result-conf { font-size: 16px !important; text-align: center !important; color: #888 !important; border: none !important; background: transparent !important; }
.result-desc { font-size: 14px !important; text-align: center !important; font-style: italic !important; color: #aaa !important; border: none !important; background: transparent !important; }
footer { display: none !important; }
@media (max-width: 768px) { .result-big { font-size: 26px !important; } }
"""

with gr.Blocks(title="Nhận diện Áo & Quần AI", theme=gr.themes.Soft(), css=css) as app:
    gr.Markdown("""
    <div style='text-align:center; padding:20px 0 10px'>
        <h1 style='font-size:28px; margin:0'>👗 Nhận diện Áo hay Quần (Bản Sửa Lỗi)</h1>
        <p style='color:#888; margin:6px 0 0'>Thuật toán tự động căn chỉnh ảnh theo chuẩn dữ liệu Fashion MNIST</p>
    </div>
    """)
    with gr.Row():
        with gr.Column(scale=1, min_width=280):
            with gr.Tabs() as input_tabs:
                with gr.Tab("📁 Tải ảnh lên", id="upload_tab"):
                    upload = gr.Image(label="Ảnh trang phục", type="numpy")
                    btn    = gr.Button("🔍 Nhận diện ảnh", variant="primary", size="lg")

                with gr.Tab("✏️ Vẽ tay", id="sketch_tab"):
                    # Tăng độ dày nét cọ lên 18 để AI dễ nhìn thấy nét vẽ hơn
                    sketch = gr.Sketchpad(
                        label="Vẽ áo hoặc quần (Nên vẽ rõ ràng, tô đậm nếu cần)",
                        type="numpy",
                        layers=False,
                        brush=gr.Brush(colors=["#000000"], default_size=18)
                    )
                    btn_sketch = gr.Button("📷 Chụp & Gửi nhận diện", variant="secondary", size="lg")

            gr.Markdown("""
            > 💡 **Mẹo giúp AI nhận diện chuẩn 100%:**
            > - **Khi vẽ tay:** Hãy vẽ chiếc quần hoặc áo **bản to rộng** (đừng vẽ bằng các đường chỉ mảnh), cố gắng phác họa rõ hình dáng đặc trưng (ví dụ: quần có 2 ống dài, áo thun có 2 ống tay ngắn).
            > - **Khi tải ảnh:** Tải ảnh chụp **trực diện**, sản phẩm nằm ở trung tâm và chiếm phần lớn bức ảnh. Tránh ảnh chụp người đang mặc đứng từ xa hoặc có bối cảnh quá phức tạp.
            """)

        with gr.Column(scale=1, min_width=280):
            gr.Markdown("#### Kết quả")
            out_l = gr.Textbox(label="", elem_classes="result-big",  interactive=False, show_label=False)
            out_c = gr.Textbox(label="", elem_classes="result-conf", interactive=False, show_label=False)
            out_d = gr.Textbox(label="", elem_classes="result-desc", interactive=False, show_label=False)

    btn.click(fn=predict_fashion_mnist, inputs=upload, outputs=[out_l, out_c, out_d])

    btn_sketch.click(
        fn=send_sketch_to_upload,
        inputs=sketch,
        outputs=[upload, input_tabs, out_l, out_c, out_d]
    )

app.launch(share=True)
```

```python
import numpy as np
from keras.datasets import fashion_mnist
from keras.utils import to_categorical

# 1. Tải dữ liệu từ thư viện
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# 2. Tiền xử lý dữ liệu đầu vào (Bắt buộc cho ANN)
# Chuyển ma trận ảnh (28x28) thành vector phẳng (784) và ép kiểu dữ liệu số thực
x_train = x_train.reshape((60000, 28 * 28)).astype('float32') / 255.0
x_test = x_test.reshape((10000, 28 * 28)).astype('float32') / 255.0

# 3. Chuyển nhãn (Labels) thành dạng cấu trúc One-hot Encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test)
```

```python
from keras.models import Sequential
from keras.layers import Dense

# Khởi tạo mô hình tuần tự
model = Sequential()

# Thêm tầng ẩn thứ nhất (đồng thời khai báo 784 đầu vào)
model.add(Dense(512, activation='relu', input_shape=(28 * 28,)))

# Thêm tầng ẩn thứ hai
model.add(Dense(256, activation='relu'))

# Thêm tầng đầu ra (10 phân lớp thời trang)
model.add(Dense(10, activation='softmax'))

# Biên dịch mô hình với thuật toán tối ưu RMSprop
model.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])

# Tiến hành huấn luyện (Train) qua 10 vòng lặp
model.fit(x_train, y_train, epochs=10, batch_size=128, validation_data=(x_test, y_test))

# Lưu bộ não AI thành file .h5 để đem đi làm App
model.save('model_ao_quan_ann.h5')
print("🎉 Đã huấn luyện và lưu mô hình ANN thành công!")
```

```python
# ==========================================================
# 1. TẢI DỮ LIỆU FASHION MNIST VÀ TIỀN XỬ LÝ
# ==========================================================
from keras.datasets import fashion_mnist
from keras.utils import to_categorical
from keras.models import Sequential
from keras.layers import Dense
import numpy as np

print("⏳ Đang tải dữ liệu Fashion MNIST từ hệ thống...")
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# Ép ma trận 2D (28x28) về Vector phẳng 1D (784) cho mạng ANN
x_train = x_train.reshape((60000, 28 * 28)).astype('float32') / 255.0
x_test = x_test.reshape((10000, 28 * 28)).astype('float32') / 255.0

# Chuyển đổi nhãn sang dạng ma trận One-hot đại diện cho 10 lớp gốc
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# ==========================================================
# 2. XÂY DỰNG CẤU TRÚC MẠNG ANN PHẲNG (DENSE LAYERS)
# ==========================================================
model = Sequential()
# Tầng ẩn 1: Tiếp nhận 784 điểm ảnh đầu vào, kích hoạt Relu
model.add(Dense(512, activation='relu', input_shape=(28 * 28,)))
# Tầng ẩn 2: Rút gọn thông tin xuống 256 nơ-ron
model.add(Dense(256, activation='relu'))
# Tầng ra: 10 nơ-ron tương ứng 10 phân lớp thời trang, xuất ra xác suất %
model.add(Dense(10, activation='softmax'))

# Biên dịch mô hình với thuật toán tối ưu RMSpop
model.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])

# ==========================================================
# 3. CÂN BẰNG TRỌNG SỐ VÀ TIẾN HÀNH TRAIN
# ==========================================================
# Vì tập dữ liệu có tới 5 lớp Áo nhưng chỉ có 1 lớp Quần, ta cấu hình
# Class Weight để phạt mô hình nặng gấp 5 lần nếu nó đoán sai chiếc Quần.
# Điều này giúp giải quyết triệt để lỗi "App chỉ nhận diện ra Áo".
custom_weights = {0: 1.0, 1: 5.0, 2: 1.0, 3: 1.0, 4: 1.0, 5: 1.0, 6: 1.0, 7: 1.0, 8: 1.0, 9: 1.0}

print("\n🚀 AI bắt đầu học (Huấn luyện qua 15 vòng lặp)...")
model.fit(
    x_train, y_train,
    epochs=15,
    batch_size=128,
    validation_data=(x_test, y_test),
    class_weight=custom_weights,
    verbose=1
)

# ==========================================================
# 4. LƯU KẾT QUẢ VÀO THƯ MỤC TRÊN GOOGLE DRIVE
# ==========================================================
import os
save_dir = '/content/drive/MyDrive/nhận diện áo-quần/'

# Tự động tạo thư mục trên Drive nếu chưa có
if not os.path.exists(save_dir):
    os.makedirs(save_dir)

model_path = os.path.join(save_dir, 'model_ao_quan_v2.h5')
model.save(model_path)

print(f"\n🎉 HOÀN THÀNH VÀ THÀNH CÔNG!")
print(f"📁 File bộ não AI đã được lưu tại: {model_path}")
print("👉 Bây giờ bạn đã có thể bật ô code App Gradio ở bước trước lên để chạy được rồi!")
```

```python
import gradio as gr
import numpy as np
import tensorflow as tf
from PIL import Image

# 1. Tải mô hình đã lưu từ Drive
try:
    model_real = tf.keras.models.load_model('/content/drive/MyDrive/nhận diện áo-quần/model_ao_quan_v2.h5')
    print("🤖 Đã nạp bộ não AI thành công! Hệ thống sẵn sàng nhận diện.")
except Exception as e:
    print(f"❌ Lỗi: Không đọc được file từ Drive. Chi tiết: {str(e)}")

EMOJI = {0: '👕', 1: '👖'}
LABEL = {0: 'ÁO', 1: 'QUẦN'}

# 2. Hàm xử lý logic ảnh thô đưa vào mạng ANN
def predict_upload(upload_image):
    if upload_image is None: return "⚠️", "Vui lòng tải hoặc vẽ ảnh!", ""
    try:
        if isinstance(upload_image, dict):
            upload_image = upload_image.get("composite", upload_image.get("background", None))
        if upload_image is None: return "⚠️", "Chưa nhận được hình ảnh!", ""
        if not isinstance(upload_image, np.ndarray): upload_image = np.array(upload_image)

        img_pil = Image.fromarray(upload_image.astype('uint8'))
        if img_pil.mode == 'RGBA':
            white_bg = Image.new("RGB", img_pil.size, (255, 255, 255))
            white_bg.paste(img_pil, mask=img_pil.split()[3])
            img_pil = white_bg

        # Chuẩn hóa ảnh xám về 28x28 theo chuẩn ANN
        img_gray = img_pil.convert('L').resize((28, 28))
        img_np = np.array(img_gray)

        # Đảo màu thông minh (Ép về nét trắng nền đen)
        corner_pixels = [img_np[0,0], img_np[0,-1], img_np[-1,0], img_np[-1,-1]]
        img_inverted = 255 - img_np if np.mean(corner_pixels) > 127 else img_np

        # Biến đổi thành ma trận phẳng (1, 784)
        img_ready = img_inverted.reshape((1, 28 * 28)).astype('float32') / 255.0
        preds = model_real.predict(img_ready, verbose=0)[0]

        # Gom cụm đầu ra chống lỗi lệch nhóm
        idx = np.argmax(preds)
        if idx in [0, 2, 3, 4, 6]:
            idx = 0
            conf = (preds[0] + preds[2] + preds[3] + preds[4] + preds[6]) * 100
        elif idx == 1:
            idx = 1
            conf = preds[1] * 100
        else:
            score_ao = preds[0] + preds[2] + preds[3] + preds[4] + preds[6]
            idx = 1 if preds[1] > score_ao else 0
            conf = (preds[1] if idx == 1 else score_ao) * 100

        return f"{EMOJI[idx]}  {LABEL[idx]}", f"Độ chắc chắn: {min(conf, 100.0):.1f}%", f"Model nhận diện đây là {LABEL[idx].lower()}"
    except Exception as e:
        return "❌", f"Lỗi xử lý: {str(e)}", ""

def send_sketch_to_upload(sketch_image):
    if sketch_image is None: return None, gr.update(), "⚠️", "Hãy vẽ gì đó!", ""
    actual_img = sketch_image.get("composite", None) if isinstance(sketch_image, dict) else sketch_image
    if actual_img is None: return None, gr.update(), "⚠️", "Không tìm thấy nét vẽ!", ""
    res_l, res_c, res_d = predict_upload(actual_img)
    return actual_img, gr.update(selected="upload_tab"), res_l, res_c, res_d

# 3. Giao diện Web Gradio
css = """
.result-big { font-size: 36px !important; font-weight: 700 !important; text-align: center !important; border: none !important; background: transparent !important; }
.result-conf { font-size: 16px !important; text-align: center !important; color: #888 !important; border: none !important; background: transparent !important; }
.result-desc { font-size: 14px !important; text-align: center !important; font-style: italic !important; color: #aaa !important; border: none !important; background: transparent !important; }
footer { display: none !important; }
"""

with gr.Blocks(title="Nhận diện Áo & Quần", theme=gr.themes.Soft(), css=css) as app:
    gr.Markdown("<div style='text-align:center; padding:20px 0 10px'><h1 style='font-size:28px; margin:0'>👗 Nhận diện Áo hay Quần</h1><p style='color:#888; margin:6px 0 0'>Chuẩn cấu trúc ANN phẳng</p></div>")
    with gr.Row():
        with gr.Column():
            with gr.Tabs() as input_tabs:
                with gr.Tab("📁 Tải ảnh lên", id="upload_tab"):
                    upload = gr.Image(label="Ảnh trang phục", type="numpy")
                    btn = gr.Button("🔍 Nhận diện ảnh", variant="primary", size="lg")
                with gr.Tab("✏️ Vẽ tay", id="sketch_tab"):
                    sketch = gr.Sketchpad(label="Vẽ áo/quần tại đây", type="numpy", layers=False, brush=gr.Brush(colors=["#000000"], default_size=15))
                    btn_sketch = gr.Button("📷 Chụp & Gửi nhận diện", variant="secondary", size="lg")
        with gr.Column():
            gr.Markdown("#### Kết quả")
            out_l = gr.Textbox(elem_classes="result-big", show_label=False, interactive=False)
            out_c = gr.Textbox(elem_classes="result-conf", show_label=False, interactive=False)
            out_d = gr.Textbox(elem_classes="result-desc", show_label=False, interactive=False)

    btn.click(fn=predict_upload, inputs=upload, outputs=[out_l, out_c, out_d])
    btn_sketch.click(fn=send_sketch_to_upload, inputs=sketch, outputs=[upload, input_tabs, out_l, out_c, out_d])

app.launch(share=True)
```

```python
import os
import numpy as np
import tensorflow as tf
from keras.models import Sequential
from keras.layers import Dense, LeakyReLU
from keras.utils import to_categorical
from PIL import Image
from google.colab import drive


print("⏳ Đang kết nối Google Drive...")
drive.mount("/content/drive", force_remount=True)


base_dir = '/content/drive/MyDrive/nhận diện áo-quần/'
folder_ao = os.path.join(base_dir, 'áo')
folder_quan = os.path.join(base_dir, 'file quần')

data = []
labels = []

def load_images_from_folder(folder, class_label):
    count = 0
    if not os.path.exists(folder):
        print(f"❌ Không tìm thấy thư mục: {folder}")
        return
    for filename in os.listdir(folder):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.webp')):
            try:
                img_path = os.path.join(folder, filename)
                img = Image.open(img_path).convert('L').resize((28, 28))
                img_np = np.array(img)

                corner_pixels = [img_np[0,0], img_np[0,-1], img_np[-1,0], img_np[-1,-1]]
                img_ready = 255 - img_np if np.mean(corner_pixels) > 127 else img_np

                data.append(img_ready.flatten() / 255.0)
                labels.append(class_label)
                count += 1
            except:
                continue
    print(f"✅ Đã nạp {count} ảnh từ thư mục: {os.path.basename(folder)}")


print("⏳ Đang quét dữ liệu từ Drive...")
load_images_from_folder(folder_ao, class_label=0)   # ÁO = 0
load_images_from_folder(folder_quan, class_label=1) # QUẦN = 1

X = np.array(data, dtype='float32')
y = np.array(labels, dtype='int32')

if len(X) == 0:
    print("❌ LỖI: Không tìm thấy ảnh nào trong 2 folder 'áo' và 'file quần'!")
else:

    Y = to_categorical(y, num_classes=2)


    indices = np.arange(X.shape[0])
    np.random.shuffle(indices)
    X, Y = X[indices], Y[indices]


    model = Sequential([
        Dense(512, input_shape=(28 * 28,)),
        LeakyReLU(alpha=0.1),
        Dense(256),
        LeakyReLU(alpha=0.1),
        Dense(2, activation='softmax')
    ])

    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])


    print(f"\n🚀 AI bắt đầu học từ {len(X)} ảnh tự cung cấp...")
    model.fit(X, Y, epochs=15, batch_size=16, validation_split=0.2, verbose=1)


    if not os.path.exists(base_dir): os.makedirs(base_dir)
    model_path = os.path.join(base_dir, 'model_ao_quan_v2.h5')
    model.save(model_path)
    print(f"\n🎉 HOÀN THÀNH TRAIN! Bộ não AI đã lưu tại: {model_path}")
```
