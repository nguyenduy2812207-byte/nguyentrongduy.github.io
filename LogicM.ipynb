```python
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import numpy as np

food=ctrl.Antecedent(np.arange(0,10.1,0.1),'food')
service=ctrl.Antecedent(np.arange(0,10.1,0.1),'service')
tip=ctrl.Consequent(np.arange(10,25.1,0.1),'tip')

```

```python
food['poor']=fuzz.trimf(food.universe,[0,0,5])
food['normal']=fuzz.trimf(food.universe,[0,5,10])
food['good']=fuzz.trimf(food.universe,[5,10,10])
service['poor']=fuzz.trimf(service.universe,[0,0,5])
service['normal']=fuzz.trimf(service.universe,[0,5,10])
service['good']=fuzz.trimf(service.universe,[5,10,10])
tip['less']=fuzz.trimf(tip.universe,[10,13.25,17.5])
tip['normal']=fuzz.trimf(tip.universe,[13.25,17.5,21.25])
tip['much']=fuzz.trimf(tip.universe,[17.5,21.25,25])
```

```python
food.view()
service.view()
tip.view()
```

```python
rule1=ctrl.Rule(food['poor']&service['poor'],tip['less'])
rule2=ctrl.Rule(food['poor']&service['normal'],tip['less'])
rule3=ctrl.Rule(food['poor']&service['good'],tip['normal'])
rule4=ctrl.Rule(food['normal']&service['poor'],tip['less'])
rule5=ctrl.Rule(food['normal']&service['normal'],tip['normal'])
rule6=ctrl.Rule(food['normal']&service['good'],tip['much'])
rule7=ctrl.Rule(food['good']&service['poor'],tip['normal'])
rule8=ctrl.Rule(food['good']&service['normal'],tip['much'])
rule9=ctrl.Rule(food['good']&service['good'],tip['much'])
```

```python
tipping_ctrl=ctrl.ControlSystem([rule1,rule2,rule3,rule4,rule5,rule6,rule7,rule8,rule9])
tipping=ctrl.ControlSystemSimulation(tipping_ctrl)
tipping.input['food']=10
tipping.input['service']=10
tipping.compute()
print(tipping.output['tip'])
tip.view(sim=tipping)
```

```python
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import numpy as np

distance=ctrl.Antecedent(np.arange(0,50,0.1),'distance')
traffic_condition=ctrl.Antecedent(np.arange(0,100,0.1),'traffic_condition')
demand_level=ctrl.Antecedent(np.arange(0,100,0.1),'demand_level')
weather_condition=ctrl.Antecedent(np.arange(0,6,0.1),'weather_condition')
customer_rating=ctrl.Antecedent(np.arange(1,5,0.1),'customer_rating')
ride_punctuality=ctrl.Antecedent(np.arange(0,100,0.1),'ride_punctuality')
cost=ctrl.Consequent(np.arange(0,10,0.1),'cost')
points=ctrl.Consequent(np.arange(0,10,0.1),'points')
```

```python
distance['short']=fuzz.trimf(distance.universe,[0,0,3])
distance['medium']=fuzz.trimf(distance.universe,[2,5,8])
distance['long']=fuzz.trimf(distance.universe,[6,13,20])
distance['very_long']=fuzz.trimf(distance.universe,[15,50,50])
traffic_condition['low']=fuzz.trimf(traffic_condition.universe,[0,0,30])
traffic_condition['medium']=fuzz.trimf(traffic_condition.universe,[20,45,70])
traffic_condition['high']=fuzz.trimf(traffic_condition.universe,[60,100,100])
demand_level['low']=fuzz.trimf(demand_level.universe,[0,0,30])
demand_level['medium']=fuzz.trimf(demand_level.universe,[20,45,70])
demand_level['high']=fuzz.trimf(demand_level.universe,[60,100,100])
weather_condition['bad']=fuzz.trimf(weather_condition.universe,[0,0,3])
weather_condition['moderate']=fuzz.trimf(weather_condition.universe,[0,3,6])
weather_condition['good']=fuzz.trimf(weather_condition.universe,[3,6,6])
customer_rating['poor']=fuzz.trimf(customer_rating.universe,[1,1,2.5])
customer_rating['average']=fuzz.trimf(customer_rating.universe,[2,3,4])
customer_rating['good']=fuzz.trimf(customer_rating.universe,[3.5,5,5])
ride_punctuality['late']=fuzz.trimf(ride_punctuality.universe,[0,0,50])
ride_punctuality['on_time']=fuzz.trimf(ride_punctuality.universe,[40,60,80])
ride_punctuality['early']=fuzz.trimf(ride_punctuality.universe,[70,100,100])
cost['low']=fuzz.trimf(cost.universe,[0,0,3])
cost['medium']=fuzz.trimf(cost.universe,[2,4,6])
cost['high']=fuzz.trimf(cost.universe,[5,7,9])
cost['very_high']=fuzz.trimf(cost.universe,[8,10,10])
points['none']=fuzz.trimf(points.universe,[0,0,0.6])
points['few']=fuzz.trimf(points.universe,[0.5,1,2])
points['moderate']=fuzz.trimf(points.universe,[1.5,4,6])
points['high']=fuzz.trimf(points.universe,[5.5,10,10])
```

```python
distance.view()
traffic_condition.view()
demand_level.view()
weather_condition.view()
customer_rating.view()
ride_punctuality.view()
cost.view()
points.view()
```

```python
rule1_cost=ctrl.Rule(distance['short']&traffic_condition['low']&demand_level['low'],cost['low'])
rule2_cost=ctrl.Rule(distance['short']&traffic_condition['medium']&demand_level['high'],cost['medium'])
rule3_cost=ctrl.Rule(distance['medium']&traffic_condition['high']&demand_level['high'],cost['high'])
rule4_cost=ctrl.Rule(distance['long']&traffic_condition['medium']&weather_condition['good'],cost['medium'])
rule5_cost=ctrl.Rule(distance['long']&traffic_condition['high']&weather_condition['bad'],cost['very_high'])
rule6_cost=ctrl.Rule(distance['very_long']&traffic_condition['high']&demand_level['high'],cost['very_high'])
rule7_cost=ctrl.Rule(distance['medium']&traffic_condition['low']&demand_level['low'],cost['medium'])
rule8_cost=ctrl.Rule(distance['short']&traffic_condition['high']&weather_condition['bad'],cost['high'])
rule9_cost=ctrl.Rule(distance['very_long']&weather_condition['bad'],cost['very_high'])
rule10_cost=ctrl.Rule(distance['medium']&traffic_condition['medium']&weather_condition['moderate'],cost['medium'])
rule11_points=ctrl.Rule(customer_rating['good']&ride_punctuality['early'],points['high'])
rule12_points=ctrl.Rule(customer_rating['average']&ride_punctuality['on_time'],points['moderate'])
rule13_points=ctrl.Rule(customer_rating['poor']&ride_punctuality['late'],points['none'])
rule14_points=ctrl.Rule(distance['long']&traffic_condition['high']&ride_punctuality['on_time'],points['high'])
rule15_points=ctrl.Rule(distance['medium']&traffic_condition['medium']&customer_rating['good'],points['moderate'])
rule16_points=ctrl.Rule(customer_rating['poor']&ride_punctuality['late'],points['none'])
rule17_points=ctrl.Rule(distance['very_long']&weather_condition['bad']&customer_rating['good'],points['high'])
rule18_points=ctrl.Rule(distance['short']&customer_rating['average']&ride_punctuality['on_time'],points['few'])
rule19_points=ctrl.Rule(distance['long']&traffic_condition['high']&ride_punctuality['late'],points['few'])
rule20_points=ctrl.Rule(distance['medium']&weather_condition['moderate']&customer_rating['good'],points['moderate'])
```

```python
tinh_cost=ctrl.ControlSystem([rule1_cost,rule2_cost,rule3_cost,rule4_cost,rule5_cost,rule6_cost,rule7_cost,rule8_cost,rule9_cost,rule10_cost])
tinh_points=ctrl.ControlSystem([rule11_points,rule12_points,rule13_points,rule14_points,rule15_points,rule16_points,rule17_points,rule18_points,rule19_points,rule20_points])

tinh_cost_sim = ctrl.ControlSystemSimulation(tinh_cost)
tinh_points_sim = ctrl.ControlSystemSimulation(tinh_points)

tinh_cost_sim.input['distance'] = 2.5
tinh_cost_sim.input['traffic_condition'] = 25
tinh_cost_sim.input['demand_level'] = 25
tinh_cost_sim.input['weather_condition'] = 3
tinh_cost_sim.compute()

tinh_points_sim.input['distance'] = 8
tinh_points_sim.input['traffic_condition'] = 90
tinh_points_sim.input['ride_punctuality'] = 40
tinh_points_sim.input['customer_rating'] = 3.5
tinh_points_sim.input['weather_condition'] = 3
tinh_points_sim.compute()

print('Cost:', tinh_cost_sim.output['cost'], 'Points:', tinh_points_sim.output['points'])
cost.view(sim=tinh_cost_sim)
points.view(sim=tinh_points_sim)
```

```python
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import numpy as np

order_density=ctrl.Antecedent(np.arange(0,6.1,0.1),'order_density')
delivery_urgency=ctrl.Antecedent(np.arange(0,6.1,0.1), 'delivery_urgency')
driver_current_load=ctrl.Antecedent(np.arange(0,6.1,0.1),'driver_current_load')
traffic_conditions=ctrl.Antecedent(np.arange(0,6.1,0.1),'traffic_conditions')
profit_per_delivery=ctrl.Antecedent(np.arange(0,6.1,0.1),'profit_per_delivery')
n=ctrl.Consequent(np.arange(0,6.1,0.1),'n')
d=ctrl.Consequent(np.arange(0,6.1,0.1),'d')
```

```python
order_density['low']=fuzz.trimf(order_density.universe,[0,0,3])
order_density['medium']=fuzz.trimf(order_density.universe,[0,3,6])
order_density['high']=fuzz.trimf(order_density.universe,[3,6,6])
delivery_urgency['low']=fuzz.trimf(delivery_urgency.universe,[0,0,3])
delivery_urgency['medium']=fuzz.trimf(delivery_urgency.universe,[0,3,6])
delivery_urgency['high']=fuzz.trimf(delivery_urgency.universe,[3,6,6])
driver_current_load['low']=fuzz.trimf(driver_current_load.universe,[0,0,3])
driver_current_load['medium']=fuzz.trimf(driver_current_load.universe,[0,3,6])
driver_current_load['high']=fuzz.trimf(driver_current_load.universe,[3,6,6])
traffic_conditions['low']=fuzz.trimf(traffic_conditions.universe,[0,0,3])
traffic_conditions['medium']=fuzz.trimf(traffic_conditions.universe,[0,3,6])
traffic_conditions['high']=fuzz.trimf(traffic_conditions.universe,[3,6,6])
profit_per_delivery['low']=fuzz.trimf(profit_per_delivery.universe,[0,0,3])
profit_per_delivery['medium']=fuzz.trimf(profit_per_delivery.universe,[0,3,6])
profit_per_delivery['high']=fuzz.trimf(profit_per_delivery.universe,[3,6,6])
n['low']=fuzz.trimf(n.universe,[0,1.5,3])
n['medium']=fuzz.trimf(n.universe,[1.5,3,4.5])
n['high']=fuzz.trimf(n.universe,[3,4.5,6])
d['low']=fuzz.trimf(d.universe,[0,1.5,3])
d['medium']=fuzz.trimf(d.universe,[1.5,3,4.5])
d['high']=fuzz.trimf(d.universe,[3,4.5,6])
```
