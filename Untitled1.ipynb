```python
import osmnx as ox
import networkx as nx
import matplotlib.pyplot as plt
import time

place_name = "District 1, Ho Chi Minh City, Vietnam"
graph = ox.graph_from_place(place_name, network_type='drive')

nodes = list(graph.nodes)
orig_node = nodes[0]
dest_node = nodes[len(nodes) // 2]

start_time = time.time()
route_dijkstra = nx.shortest_path(graph, orig_node, dest_node, weight='length', method='dijkstra')
dijkstra_time = time.time() - start_time

start_time = time.time()
route_astar = nx.astar_path(graph, orig_node, dest_node, weight='length')
astar_time = time.time() - start_time

fig, ax = ox.plot_graph_route(graph, route_dijkstra, route_color='blue', route_linewidth=5, node_size=0)
print(f"Thời gian chạy Dijkstra: {dijkstra_time:.6f} giây")
print(f"Thời gian chạy A*: {astar_time:.6f} giây")
```

```python
!pip install osmnx
import osmnx as ox
import networkx as nx
import matplotlib.pyplot as plt
import random

place_name = "District 1, Ho Chi Minh City, Vietnam"
graph = ox.graph_from_place(place_name, network_type='drive')
nodes_list = list(graph.nodes)

customer_node = random.choice(nodes_list)
vehicle_nodes = [random.choice(nodes_list) for _ in range(5)]

def find_best_vehicle(graph, customer, vehicles):
    best_vehicle = None
    min_dist = float('inf')

    for vehicle in vehicles:
        try:
            dist = nx.shortest_path_length(graph, vehicle, customer, weight='length')
            if dist < min_dist:
                min_dist = dist
                best_vehicle = vehicle
        except nx.NetworkXNoPath:
            continue
    return best_vehicle, min_dist

best_v, distance = find_best_vehicle(graph, customer_node, vehicle_nodes)

route = nx.shortest_path(graph, best_v, customer_node, weight='length')

print(f"Khoảng cách đến xe gần nhất: {distance:.2f} mét")

fig, ax = ox.plot_graph_route(
    graph, route, route_color='green', route_linewidth=6, node_size=0, show=False, close=False
)
customer_coords = (graph.nodes[customer_node]['y'], graph.nodes[customer_node]['x'])
ax.scatter(customer_coords[1], customer_coords[0], c='red', s=200, marker='*', label='Khách hàng', zorder=5)

for v in vehicle_nodes:
    v_coords = (graph.nodes[v]['y'], graph.nodes[v]['x'])
    color = 'yellow' if v == best_v else 'blue'
    ax.scatter(v_coords[1], v_coords[0], c=color, s=100, marker='o', edgecolors='black', zorder=4)

plt.legend()
plt.show()
```
